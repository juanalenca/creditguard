# Análise Exploratória de Dados (EDA) - CreditGuard AI
Neste notebook, geraremos e analisaremos um dataset simulado (mock) contendo clientes e contratos para cálculo de KPIs base.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from faker import Faker

# Configurações estéticas
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Geração de Dados Simulados (Mocking)

```python
fake = Faker('pt_BR')
Faker.seed(42)
np.random.seed(42)

num_clientes = 150
regioes = ['Nordeste', 'Sudeste']
status_contrato = ['ativo', 'inadimplente', 'quitado']

dados = []
for _ in range(num_clientes):
    regiao = np.random.choice(regioes)
    saldo_devedor = round(np.random.uniform(1000, 50000), 2)
    status = np.random.choice(status_contrato, p=[0.7, 0.2, 0.1])
    atraso_dias = np.random.randint(0, 120) if status == 'inadimplente' else 0
    
    dados.append({
        'regiao': regiao,
        'saldo_devedor': saldo_devedor,
        'status': status,
        'atraso_dias': atraso_dias
    })

df = pd.DataFrame(dados)
df.head()
```

## 2. Limpeza e Tratamento Básicos (Data Cleaning)

```python
# Verificar valores nulos
print("Valores nulos por coluna:\n", df.isnull().sum())

# Classificar nível de risco baseado em dias de atraso (Regra de Negócio)
df['nivel_risco'] = df['atraso_dias'].apply(
    lambda x: 'Alto' if x > 60 else ('Medio' if x > 15 else 'Baixo')
)
df[['atraso_dias', 'nivel_risco']].value_counts()
```

## 3. Análise Visual de Inadimplência

```python
inadimplentes = df[df['status'] == 'inadimplente']

plt.title('Distribuição de Dias de Atraso (Inadimplentes)')
sns.histplot(inadimplentes['atraso_dias'], bins=15, kde=True, color='red')
plt.xlabel('Dias de Atraso')
plt.ylabel('Contagem de Contratos')
plt.show()
```

## 4. Cálculo Simulados dos KPIs

```python
inadimplencia_total = df[df['status'] == 'inadimplente']['saldo_devedor'].sum()
atraso_medio = df[df['status'] == 'inadimplente']['atraso_dias'].mean()
clientes_criticos = df[df['nivel_risco'] == 'Alto'].shape[0]

print(f"KPI - Inadimplência Total: R$ {inadimplencia_total:,.2f}")
print(f"KPI - Atraso Médio: {atraso_medio:.1f} dias")
print(f"KPI - Clientes Críticos (Risco Alto): {clientes_criticos}")
```
